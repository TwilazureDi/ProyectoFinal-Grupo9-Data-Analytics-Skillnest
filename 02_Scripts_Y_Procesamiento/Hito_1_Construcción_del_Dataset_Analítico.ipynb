## **¿Qué tipo de análisis permitirá tu dataset en el siguiente módulo?**

Observar las tendencias de los clientes y su segmentacion sobre su localizacion, rentabilidad, preferencia de productos, categorias y marcas.
Cada registro corresponde a la compra individual de un producto.

----

## **¿Qué tablas decidiste utilizar y por qué?**

Las tablas utilizadas fueron: `Clientes`, `Productos`, `Geografia`,`Clientes_Ubicacion`,`Pedidos`.

Nuestro objetivo es analizar el comportamiento y tendencias de cada `Segmentacion` de la `Tabla Clientes`, relacionado a su ubicacion y los productos que compran.

----

## **¿Qué columnas seleccionaste y cuáles descartaste?**

#### **Columnas Descartadas:**

`ID_Campania`,`ID_Producto`,`Codigo_Barra`,`Codigo_Interno`,`ID_Ubicacion`,`ID_Detalle`:
Son variables de referencia y llaves una vez limpiado en dataframe no tienen valor para el analisis.

--

`Precio_Lista`,`Costo_Unitario`,`Descuento`,`Fecha_Envio`,`Zona_Logistica`,`Activo`, `Detalle_Pedido`:
Son variables irrelevantes para nuestro analisis enfocado observar el comportamiento y tendencias de los clientes.

--

`Pais`,`Tipo_Direccion`:
Todo el registro indica que son solo las sucursales de Chile, asi como solo estan registrado los tipo de direccion principales.

--

`Codigo_Postal`,`Latitud_Ref`:
Una direccion demasiado especifica no es requerida para este tipo de analisis.



#### **Columnas Seleccionadas:**
`ID_Cliente`,`Nombre_Cliente`,`Fecha_Registro`,`Antiguedad_Dias`,`Segmento`,
`Puntaje_Fidelidad`,`Canal_Preferido`y  `Fecha_Actualizacion`: Casi todos de la columna de la tabla `Clientes`, nuestro principal foco de analisis sobre su comportamiento. Con `Antiguedad_Dias` nuestra variable calculada. Y con `ID_Clientes` permitiendonos calcular la cantidad de Clientes registrados.

--

`Region`,`Estado`,`Ciudad`: Columnas de las tabla de `Geografia` utilizando `Clientes_Ubicacion` como intermedio para obtener la ubicacion aproximada de los Clientes.

--

`ID_Pedido`,`Fecha_Pedido`,`Modo_Envio`,`Prioridad`,`Estado_Pedido`, `Etiqueta_Ventas` y `Etiqueta_Ganancia`,`Ventas`,`Ganancia`,`Margen_Rentabilidad_%`: Columnas de la tabla `Pedido`, utilizado para evaluar tendencias y rentabilidad de los clientes y segmentacion. Siendo `Margen_Rentabilidad_%` nuestra otra variable calculada. Y con `ID_Pedido` permitiendonos calcular la cantidad de pedidos realizados.

--

`Nombre_Producto`,`Categoria`,`Subcategoria`,`Marca`,`Cantidad`: Columnas de la tabla `Productos` habiendo utilizado de intermedio la tabla `Detalle_Pedido`, utilizado para evaluar tendencias y preferencias de los clientes respecto a los productos, categorias y marcas mas pedidos, asi como cuales son los segmentos que mas cantidades compran.

### Tabla detalle pedidos

```python
print("--- 1. ANÁLISIS DE DATOS NULOS O FALTANTES ---")
print(df_det_pedidos.isnull().sum())

print("\n--- 2. DISTRIBUCIÓN DE VOLUMEN (CANTIDADES) ---")
print(df_det_pedidos['Cantidad'].value_counts().sort_index())
```

```python
# 1. Calcular IQR para Ganancias y Ventas
Q1_g = df_det_pedidos['Ganancia'].quantile(0.25)
Q3_g = df_det_pedidos['Ganancia'].quantile(0.75)
IQR_g = Q3_g - Q1_g
limite_inf_g = Q1_g - 1.5 * IQR_g
limite_sup_g = Q3_g + 1.5 * IQR_g

Q3_v = df_det_pedidos['Ventas'].quantile(0.75)
limite_sup_v = Q3_v + 1.5 * (Q3_v - df_det_pedidos['Ventas'].quantile(0.25))

# 2. Crear las variables calculadas por defecto
df_det_pedidos['Etiqueta_Ganancia'] = 'Normal'
df_det_pedidos['Etiqueta_Ventas'] = 'Normal'

# 3. Aplicar las etiquetas a los outliers detectados sean positivos o negativos.
df_det_pedidos.loc[df_det_pedidos['Ganancia'] > limite_sup_g, 'Etiqueta_Ganancia'] = 'Ganancia Atípica'
df_det_pedidos.loc[df_det_pedidos['Ganancia'] < limite_inf_g, 'Etiqueta_Ganancia'] = 'Pérdida Crítica'
df_det_pedidos.loc[df_det_pedidos['Ventas'] > limite_sup_v, 'Etiqueta_Ventas'] = 'Venta Extraordinaria'

print("Variables calculadas agregadas exitosamente.")
```

```python
# 4. Verificación de duplicados
print(df_det_pedidos.duplicated().sum())
```

```python
# 5. Buscar por si hay IDs fantasmas cruzando con las tablas maestras
pedidos_fantasma = df_det_pedidos[~df_det_pedidos['ID_Pedido'].isin(df_pedidos['ID_Pedido'])]
productos_fantasma = df_det_pedidos[~df_det_pedidos['ID_Producto'].isin(df_productos['ID_Producto'])]

print("--- RESULTADOS DE LA VALIDACIÓN DE INTEGRIDAD ---")
print(f"Fallas en Pedidos: {len(pedidos_fantasma)} registros huérfanos.")
print(f"Fallas en Productos: {len(productos_fantasma)} registros huérfanos.")
```

```python
# Creación de métrica financiera, variable calculada numérica.
df_det_pedidos['Margen_Rentabilidad_%'] = round((df_det_pedidos['Ganancia'] / df_det_pedidos['Ventas']) * 100, 2)
```

## Joins de Tablas

```python
#Joins de las tablas de ubicacion
df_clientes_Ubi_1 = pd.merge(df_clientes, df_cli_ubicacion, on= 'ID_Cliente', how='left')

df_clientes_Ubi_Complete = pd.merge(df_clientes_Ubi_1, df_geografia, on = 'ID_Ubicacion', how= 'left')
df_clientes_Ubi_Complete.info()
```

```python
#Joins de las tablas de pedidos
df_pedidos_detPedidos = pd.merge(df_pedidos, df_det_pedidos, on= 'ID_Pedido', how= 'left' )

df_pedidos_Complete = pd.merge(df_pedidos_detPedidos, df_productos, on= 'ID_Producto', how= 'left')
df_pedidos_Complete.info()
```

```python
#Join de la tabla final y verificacion
df_clientes_Final = pd.merge(df_pedidos_Complete, df_clientes_Ubi_Complete, on='ID_Cliente', how ='left')
df_clientes_Final.info()
```

```python
#Limpieza de clientes Final
df_clientes_Final = df_clientes_Final.drop(columns=['ID_Campania', 'ID_Producto','ID_Detalle',
                                                    'Precio_Lista','Costo_Unitario','Codigo_Barra','Codigo_Interno','Ciudad',
                                                    'ID_Ubicacion','Codigo_Postal','Zona_Logistica', 'Tipo_Direccion',
                                                    'Pais','Descuento', 'Fecha_Envio','Activo','Latitud_Ref','Observacion_Interna'])



```

```python
df_clientes_Final = df_clientes_Final[[ 'ID_Cliente','Nombre_Cliente','Fecha_Registro','Antiguedad_Dias','Segmento',
                                       'Puntaje_Fidelidad','Canal_Preferido','Region','Estado','Fecha_Actualizacion','ID_Pedido',
                                       'Fecha_Pedido','Modo_Envio','Prioridad','Estado_Pedido','Ventas','Etiqueta_Ventas','Ganancia','Etiqueta_Ganancia',
                                       'Margen_Rentabilidad_%','Nombre_Producto','Cantidad','Categoria','Subcategoria','Marca'
                                       ]]
```

```python
df_clientes_Final['Antiguedad_Dias'] = (df_clientes_Final['Fecha_Pedido'] - df_clientes_Final['Fecha_Registro']).dt.days
```
